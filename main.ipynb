```python
# CSC 348 Homework 2 
import numpy as np 
import matplotlib.pyplot as plt 
```

```python
# a. Make a 10x10 diagonal matrix with three 1s on the diagonal, and all other entries zero 
A = np.zeros((10, 10), dtype=int)

A[0][0] = 1
A[1][1] = 1
A[2][2] = 1

print(A) 
```

```python
# b. what is the rank of this matrix? Find the eigenvalues and eigenvectors. 
rank = np.linalg.matrix_rank(A)
eigenvalues, eigenvectors = np.linalg.eig(A)

print(f"rank: {rank}\n eigenvalues: {eigenvalues}\n eigenvectors: {eigenvectors}")
```

```python
# c. Now add a small random number to each entry of size. What is the rank of this matrix? How do the eigenvalues and eigenvectors change?
noise = np.random.uniform(low=-0.01, high=0.01, size=(10, 10))
A_noisy = A + noise

print(f"Original Rank: {np.linalg.matrix_rank(A)}")
print(f"New Rank: {np.linalg.matrix_rank(A_noisy)}")

eigenvalues, eigenvectors = np.linalg.eig(A_noisy)
print("\nNew Eigenvalues:")
print(np.round(eigenvalues, 4))
print(eigenvectors)

"""
The rank goes from 3 --> 10. This is expected becaue a small entry is now added onto each entry of the original A.
I think the interesting takeaway from this is for very sparse matricies approximations are great b/c the only values that really show any 
important information
"""
```

```python
Z = np.zeros((10, 10))
noise = np.random.uniform(low=-0.01, high=0.01, size=(10, 10))
Z = Z + noise

print(f"Zero Matrix + Noise Rank: {np.linalg.matrix_rank(Z)}")

eigenvalues, eigenvectors = np.linalg.eig(Z)
print("\nNew Eigenvalues:")
print(np.round(eigenvalues, 4))

print("Eigenvectors:")
print(eigenvectors)

""" 
B/c all of these noisy values are very small numbers from [-0.01, 0.01] the computer doesn't actually recongize them as values here
and the rank is 0 which is a interesting change from having rank 10 earlier
"""
```

```python
coeffs = [1, -18, 144, -672, 2016, -4032, 5376, -4608, 2304, -512]

x = np.linspace(1.920, 2.080, 100)

y_expanded = np.polyval(coeffs, x)

y_factored = (x - 2)**9

plt.figure(figsize=(10, 6))

plt.plot(x, y_expanded, label='Expanded Form', color='red', alpha=0.8)

plt.plot(x, y_factored, label='Factored Form: $(x-2)^9$', color='black', linewidth=2)

plt.title("Evaluating $p(x) = (x-2)^9$ near $x=2$")
plt.xlabel("x")
plt.ylabel("p(x)")
plt.legend()
plt.grid(True)
plt.show()

""" 
In the expanded form there seems to be a lot of noise because of exact calculations made on small floating point integers.
For instance, evaluating the coefficient for the expanded form for lets say p(1.926) = (1.926)^6 - 18(1.928)^8 ... will cause a lot of floating 
point imprecison added up in comparsion to just p(1.926) = (1.926 - 2)^9.
"""
test = np.polyval(coeffs, 1.921)
print(test)
```

```python
def polynomial(x):
    num = (
        (x**9)
        - (18 * (x**8))
        + (144 * (x**7))
        - (672 * (x**6))
        + (2016 * (x**5))
        - (4032 * (x**4))
        + (5376 * (x**3))
        - (4608 * (x**2))
        + (2304 * (x))
        - 512
    )
    return num


polynomial(1.92)
```
